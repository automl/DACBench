```python
from __future__ import annotations

from pathlib import Path
```

```python
def write_instances(filename, size, init_objs):
    """Write an instance set with schema id,size,initObj."""
    lines = ["id,size,initObj"]
    lines += [f"{i},{size},{init_obj}" for i, init_obj in enumerate(init_objs)]
    Path(filename).write_text("\n".join(lines) + "\n")
```

```python
# OneMax: initObj ladder {random, 0, n//4, n//2, n-1}, plus easy / random sets
for n in [20, 50, 100, 200]:
    write_instances(f"onemax_{n}.csv", n, ["random", 0, n // 4, n // 2, n - 1])
for n in [20, 50, 100]:
    write_instances(f"onemax_{n}_easy.csv", n, [n - 1])
    write_instances(f"onemax_{n}_random.csv", n, ["random"])
```

```python
# OneMax prior-art sizes (OneMax-DAC-Extension, om_ollga_{n}_medium.csv: a
# fraction 0.5 of one-bits, converted to an integer count n//2), plus a random row
for n in [300, 500, 1000, 2000]:
    write_instances(f"onemax_{n}.csv", n, ["random", n // 2])
```

```python
# LeadingOnes: same initObj ladders as instance_sets/theory/lo_rls_{n}.csv
lo_ladders = {
    50: ["random", 10, 20, 40, 45, 49],
    100: ["random", 20, 50, 70, 90, 95],
    200: ["random", 50, 100, 150, 190, 195],
}
for n, init_objs in lo_ladders.items():
    write_instances(f"leadingones_{n}.csv", n, init_objs)
```
